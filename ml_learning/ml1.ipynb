```python
import pandas as pd

data = {
    "площадь":     [38, 45, 60, 55, 80, 22, 110, 70],      # м²
    "комнат":      [1,  2,  2,  2,  3,  1,  4,   3],
    "район":       ["центр", "север", "юг", "север", "центр", "юг", "центр", "север"],
    "балкон":      [0,  1,  0,  1,  1,  0,  1,   1],        # 1 = есть, 0 = нет
    "цена_за_м2":  [320, 180, 150, 190, 350, 140, 400, 210], # тыс. руб.
}


df = pd.DataFrame(data) # превращает словарь в таблицу
df # это вывод кстати (в jupyter это специальная фишка если напистаь вот так то будет таблица красивая)
```

```python
df.info()       # типы колонок, сколько непустых значений, память
df.describe()   # статистика по числовым колонкам: count, mean, std, min, max...
df.head(3)      # первые 3 строки
df.shape        # (8, 5) — кортеж (строк, колонок)
df.columns      # имена колонок
```

```python
# В реальных данных (и в твоей лабе df_gen тоже) бывают пропуски 
# в pandas они обозначаются как NaN (Not a Number). Модели на таких данных падают, 
# поэтому пропуски надо заполнить перед обучением.

# Сначала испортим наш датасет — добавим пропуски руками, чтобы было на чём тренироваться.


# Возьмём копию df из урока 1 и испортим несколько ячеек
df_bad = df.copy()

# В pandas NaN удобно создавать через float("nan") или через None.
# В колонку "площадь" на строки 1 и 5 кладём NaN
df_bad.loc[1, "площадь"] = None
df_bad.loc[5, "площадь"] = None

# В колонку "район" на строки 2 и 7 — тоже NaN
df_bad.loc[2, "район"] = None
df_bad.loc[7, "район"] = None

df_bad
```
